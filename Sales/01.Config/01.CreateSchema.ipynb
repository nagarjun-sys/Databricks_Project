#### CREATE CATALOG,SCHEMA, VOLUME

- SCHEMA = `LANDING, BRONZE, SILVER, GOLD`

```python
spark.sql('CREATE SCHEMA IF NOT EXISTS sales.landing');
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS sales.bronze;
```
