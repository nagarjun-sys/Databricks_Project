```python
CATALOG = 'SALES'
SCHEMA = 'SILVER'
TARGET_TABLE = 'PRODUCTCATEGORYTRANSLATION_ENRICHED'
```

```python
productcategorytranslation_df = spark.table('SALES.BRONZE.PRODUCTCATEGORYTRANSLATION')
productcategorytranslation_df.limit(1).display()
```

```python
from pyspark.sql.functions import *

productcategorytranslation_enriched_df = productcategorytranslation_df.select(
    upper(trim(col('product_category_name'))).alias('product_category_name'),
    upper(trim(col('product_category_name_english'))).alias('product_category_name_english'),
    col('file_name'),
    col('file_path'),
    col('load_timestamp')
).filter('product_category_name IS NOT NULL')

display(productcategorytranslation_enriched_df.limit(1)) 
```

```python
productcategorytranslation_enriched_df.createOrReplaceTempView('enriched_productcategorytranslation_temp_vw')
```

```python
%sql
select * from enriched_productcategorytranslation_temp_vw
```

```python
spark.sql(f"""
MERGE INTO {CATALOG}.{SCHEMA}.{TARGET_TABLE} tgt
USING (
  SELECT *
  FROM (
    SELECT *,
           ROW_NUMBER() OVER (
               PARTITION BY product_category_name
               ORDER BY load_timestamp DESC
           ) rn
    FROM enriched_productcategorytranslation_temp_vw
  )
  WHERE rn = 1
) src
ON (src.product_category_name  = tgt.product_category_name)
WHEN MATCHED THEN
  UPDATE SET
    tgt.product_category_name = src.product_category_name,
    tgt.product_category_name_english = src.product_category_name_english
    
WHEN NOT MATCHED THEN
  INSERT (
    product_category_name,
    product_category_name_english,
    file_name,
    origin_file_path,
    load_timestamp
  )
  VALUES (
    src.product_category_name,
    src.product_category_name_english,
    src.file_name,
    src.file_path,
    current_timestamp()
  )
""")
```
