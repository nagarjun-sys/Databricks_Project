```python
Base_Dir = '/Volumes/sales/landing/volumes/Geolocations/'
Catalog = 'Sales'
Schema = 'bronze'
target_table = 'geolocations'

```

```python
cust_schema = ''' gelocation_zip_code_prefix string, 
geolocation_lat string, 
geolocation_lng string, 
geolocation_city string'''
```

```python


geoLoc_df = spark.read.format('csv').options(header='true').schema(cust_schema).load('/Volumes/sales/landing/volumes/Geolocations/gealocation-09.csv')
display(cust_df.limit(1))
```

```python
from pyspark.sql.functions import *
geoLoc_trans_df = geoLoc_df.select('*', '_metadata.file_name', "_metadata.file_path", '_metadata.file_modification_time', current_timestamp().alias('ingest_timestamp') )
display(geoLoc_trans_df.limit(1))
```

```python
geoLoc_df.write.mode('append').format('delta').saveAsTable(f'{Catalog}.{Schema}.{target_table}')  # Ensure catalog, schema, and target_table are defined before this line
display(geoLoc_df.limit(1))
```
