```python
Base_Dir = '/Volumes/sales/landing/volumes/OrderItems/'
Catalog = 'Sales'
Schema = 'bronze'
target_table = 'orderitemsapi'
```

```python
orderitems_schema = """order_id STRING, 
                 order_item_id INT, 
                 product_id STRING, 
                 seller_id STRING, 
                 shipping_limit_date TIMESTAMP,
                 price DECIMAL(10,2),
                 freight_value DECIMAL(10,2)
                 """
```

```python
Orderitems_df = (spark.read.format('csv').option('header','true').schema(orderitems_schema).load(f'{Base_Dir}')
           )
display(Orderitems_df.limit(3))
Orderitems_df.printSchema()
```

```python
from pyspark.sql.functions import *
Orderitems_trans_df = Orderitems_df.select('*', '_metadata.file_name', "_metadata.file_path", '_metadata.file_modification_time', current_timestamp().alias('ingest_timestamp') )
display(Orderitems_trans_df.limit(1))
```

```python
(Orderitems_trans_df.write.mode('append').format('delta')
               .saveAsTable(f'{Catalog}.{Schema}.{target_table}')
               )  # Ensure catalog, schema, and target_table are defined before this line
display(Orderitems_trans_df.limit(1))
```
