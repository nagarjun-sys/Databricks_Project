```python
Base_Dir = '/Volumes/sales/landing/volumes/Customers/'
Catalog = 'Sales'
Schema = 'bronze'
target_table = 'customersreadapi'
```

```python
customers_schema = """customer_id STRING, 
                 customer_unique_id STRING, 
                 customer_zip_code_prefix INT, 
                 customer_city STRING, 
                 customer_state STRING
                 """
```

```python
Customers_df = spark.read.format('csv').options(header='true').schema(customers_schema).load(f'{Base_Dir}')
display(Customers_df.limit(1))
```

```python
from pyspark.sql.functions import *
Customers_trans_df = Customers_df.select('*', '_metadata.file_name', "_metadata.file_path", '_metadata.file_modification_time', current_timestamp().alias('ingest_timestamp') )
display(Customers_trans_df.limit(1))
```

```python
(Customers_trans_df.write.mode('append').format('delta')
               .saveAsTable(f'{Catalog}.{Schema}.{target_table}')
               )  # Ensure catalog, schema, and target_table are defined before this line
display(Customers_trans_df.limit(1))
```
